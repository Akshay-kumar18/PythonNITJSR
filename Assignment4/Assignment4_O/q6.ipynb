```python
import matplotlib.pyplot as plt

# Data
students = ['A', 'B', 'C', 'D', 'E']
marks = [45, 67, 89, 56, 93]
student_indices = list(range(len(students)))

# 1. Histogram (Distribution of Marks)
plt.figure(figsize=(6, 4))
plt.hist(marks, bins=5, color='teal', edgecolor='black')
plt.title('Distribution of Marks')
plt.xlabel('Marks Range')
plt.ylabel('Frequency')
plt.show()

# 2. Scatter Plot (Student Index vs Marks)
plt.figure(figsize=(6, 4))
plt.scatter(student_indices, marks, color='red', s=100)
plt.title('Student Index vs Marks')
plt.xlabel('Student Index')
plt.ylabel('Marks')
plt.xticks(student_indices, labels=students)
plt.grid(True)
plt.show()

# 3. Pie Chart (Percentage Contribution to Total Marks)
plt.figure(figsize=(6, 6))
plt.pie(marks, labels=students, autopct='%1.1f%%', startangle=140)
plt.title('Student Percentage Contribution to Total Marks')
plt.show()
```
